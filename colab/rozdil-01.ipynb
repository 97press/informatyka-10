## 1.15. Ділення із заокругленням угору

### Приклад: автобуси для екскурсії

```python
%%writefile p1_15.cpp
#include <iostream>

int main() {
    int students{};
    int seatsPerBus{};

    std::cin >> students >> seatsPerBus;

    int buses = (students + seatsPerBus - 1) / seatsPerBus;
    int freeSeats = buses * seatsPerBus - students;

    std::cout << "Автобусів: " << buses << '\n';
    std::cout << "Вільних місць: " << freeSeats << '\n';

    return 0;
}
```

```python
%%writefile p1_15.txt
31 10
```

```python
!g++ -std=c++20 -Wall -Wextra p1_15.cpp -o p1_15 && ./p1_15 < p1_15.txt
```

## 1.19. Коротка шпаргалка

```python
%%writefile p1_19.cpp
#include <iostream>   // бібліотека введення-виведення

int main() {
    int a{};                       // змінна з ініціалізацією нулем
    std::cin >> a;                 // введення
    std::cout << a * 2 << '\n';    // виведення й перехід на новий рядок
    return 0;
}
```

```python
%%writefile p1_19.txt
21
```
